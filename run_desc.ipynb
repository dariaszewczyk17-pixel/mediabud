```

import subprocess, os, time

WORKDIR = "/data/workspace/919fac5a-210e-47ca-8b62-27ddea343c50/mediabud"

# Zabij stary proces
subprocess.run(["pkill", "-f", "generate_descriptions.py"], capture_output=True)
time.sleep(1)

# Wyczyść stare pliki
for f in ["desc_progress.json", "desc_run.log", "desc_log.jsonl"]:
    fp = os.path.join(WORKDIR, f)
    if os.path.exists(fp):
        os.remove(fp)
        print(f"Usunięto: {f}")

# Uruchom nową wersję v2 jako subprocess
proc = subprocess.Popen(
    ["python3", "generate_descriptions.py"],
    cwd=WORKDIR,
    stdout=open(os.path.join(WORKDIR, "desc_run.log"), "w"),
    stderr=subprocess.STDOUT,
    start_new_session=True,
)
print(f"Uruchomiono PID: {proc.pid}")

# Poczekaj 10s i sprawdź log startowy
time.sleep(10)
with open(os.path.join(WORKDIR, "desc_run.log")) as f:
    print(f.read())
```
